```python
import Bio
import py3Dmol

print("Biopython version:", Bio.__version__)
print("py3Dmol installed successfully!")
```

```python
from Bio.PDB import PDBParser, PDBIO, Select

parser = PDBParser(QUIET=True)

structure = parser.get_structure(
    "PIK3CA",
    "/content/8V8H.pdb"
)

class ChainA(Select):

    def accept_chain(self, chain):
        return chain.id == "A"

io = PDBIO()
io.set_structure(structure)

io.save(
    "/content/PIK3CA_chainA.pdb",
    ChainA()
)

print("Chain A extracted successfully!")
```

```python
import py3Dmol

with open("/content/PIK3CA_chainA.pdb") as f:
    pdb_data = f.read()

view = py3Dmol.view(width=900, height=600)

view.addModel(pdb_data, "pdb")

view.setStyle({
    "cartoon": {"color": "spectrum"}
})

view.zoomTo()
view.show()
```

```python
from Bio.PDB import PDBParser

parser = PDBParser(QUIET=True)

structure = parser.get_structure(
    "PIK3CA",
    "/content/PIK3CA_chainA.pdb"
)

for residue in structure.get_residues():

    if residue.id[0] == " ":

        if residue.id[1] == 1047:

            print("Residue:", residue.resname)
            print("Residue ID:", residue.id[1])

            for atom in residue:
                print(
                    atom.name,
                    atom.element,
                    atom.coord
                )
```

```python
view = py3Dmol.view(width=900, height=600)

view.addModel(pdb_data, "pdb")

view.setStyle({
    "cartoon": {"color": "lightgray"}
})

view.addStyle(
    {"resi": 1047},
    {"stick": {"colorscheme": "greenCarbon"}}
)

view.zoomTo({"resi": 1047})
view.show()
```
